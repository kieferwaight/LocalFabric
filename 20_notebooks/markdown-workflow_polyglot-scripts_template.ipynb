# Template Notebook: Polyglot Scripts
Template for testing polyglot script based markdown workflows

## Markdown Definition

Define markdown file as variable

````python
md_definition = """---
id: tests.polyglot
inputs:
  name:
    type: string
    required: true
---

# Polyglot

```bash {id: shell-step}
echo "shell sees {{ name }}"
ls -all
```

```python {id: py-step}
import os, json
json.dump({"py_saw": "{{ name }}"}, open(os.environ["STATE_FILE"], "w"))
```
"""
print(md_definition)
````

## Input Definitions

Define the inputs

````python
inputs = {"name": "Hello World"}

print(inputs)
````

## Run

Call execution using the template provided above

````python
from harnesses.markdown.compiler import PROVIDER_MARKER_KEY
from harnesses.markdown.harness import MarkdownHarness


class NotebookMarkdownHarness(MarkdownHarness):
    """MarkdownHarness extension that executes a markdown string directly."""

    def execute_text(self, source, arguments=None, *, source_path="<notebook>"):
        definition = self.compile_text(source, source_path=source_path)
        arguments = dict(arguments or {})
        if PROVIDER_MARKER_KEY in definition:
            return self._execute_provider(definition, arguments)
        return self._execute_via_runtime(definition, arguments)


harness = NotebookMarkdownHarness()
````

````python
result = harness.execute_text(md_definition, inputs)
print(result)
````
